```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import load_iris

# Carregar o conjunto de dados Iris
data = load_iris()
df = pd.DataFrame(data.data, columns=data.feature_names)
```

```python
# Calcular a matriz de correlação
corr_matrix = df.corr()

# Plotar a matriz de correlação como um heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Matriz de Correlação')
plt.show()
```

```python
# Plotar scatter plots para variáveis
sns.pairplot(df)
plt.suptitle('Scatter Plots das Variáveis', y=1.02)
plt.show()
```

```python
# Padronizar os dados
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)

# Executar PCA
pca = PCA()
pca_result = pca.fit_transform(scaled_data)
```

```python
# Plotar a variância explicada por cada componente
explained_variance_ratio = pca.explained_variance_ratio_

plt.figure(figsize=(10, 6))
plt.plot(range(1, len(explained_variance_ratio) + 1), explained_variance_ratio.cumsum(), marker='o')
plt.title('Curva de Variância Explicada')
plt.xlabel('Número de Componentes')
plt.ylabel('Variância Explicada Acumulada')
plt.grid()
plt.show()
```

```python
# Calcular a matriz de correlação no domínio transformado
transformed_df = pd.DataFrame(pca_result[:, :len(df.columns)], columns=[f'PC{i+1}' for i in range(len(df.columns))])
corr_matrix_transformed = transformed_df.corr()

# Plotar a matriz de correlação transformada como um heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix_transformed, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Matriz de Correlação no Domínio Transformado')
plt.show()
```

```python
# Determinar o número de componentes necessários para manter 90% da variância
cumulative_variance = explained_variance_ratio.cumsum()
n_components = np.argmax(cumulative_variance >= 0.90) + 1

print(f"Número de componentes para manter 90% da variância: {n_components}")
```

```python
# Projetar a base de dados usando os N primeiros componentes
pca = PCA(n_components=n_components)
transformed_data = pca.fit_transform(scaled_data)
reconstructed_data = pca.inverse_transform(transformed_data)

# Criar DataFrame com dados reconstruídos
reconstructed_df = pd.DataFrame(reconstructed_data, columns=df.columns)

# Comparar variáveis originais e recuperadas
comparison_df = pd.DataFrame({
    'Original': df.mean(),
    'Reconstruído': reconstructed_df.mean()
})
print(comparison_df)
```

```python
# prompt: 1. Escolha uma base de dados que represente um problema de classifica ̧c ̃ao,
# importe-a no Colab e utilizando programa ̧c ̃ao em Python realize os itens a seguir:
# a Aplique PCA, ICA e An ́alise de Discriminantes.
# b Analise os resultados obtidos plotando as PCDs e comparando o poder
# discriminativo com os componentes independentes, componentes principais e
# vari ́aveis originais.
# c Estime o resultado

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA, FastICA
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Carregar o conjunto de dados Wine
data = load_wine()
df = pd.DataFrame(data.data, columns=data.feature_names)
df['target'] = data.target

# Dividir os dados em conjuntos de treinamento e teste
X_train, X_test, y_train, y_test = train_test_split(df.drop('target', axis=1), df['target'], test_size=0.2, random_state=42)

# Padronizar os dados
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Aplicar PCA
pca = PCA(n_components=2)
X_train_pca = pca.fit_transform(X_train_scaled)
X_test_pca = pca.transform(X_test_scaled)

# Aplicar ICA
ica = FastICA(n_components=2)
X_train_ica = ica.fit_transform(X_train_scaled)
X_test_ica = ica.transform(X_test_scaled)

# Aplicar Análise Discriminante Linear (LDA)
lda = LinearDiscriminantAnalysis(n_components=2)
X_train_lda = lda.fit_transform(X_train_scaled, y_train)
X_test_lda = lda.transform(X_test_scaled)

# Plotar os resultados
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.scatter(X_train_pca[:, 0], X_train_pca[:, 1], c=y_train)
plt.title('PCA')

plt.subplot(1, 3, 2)
plt.scatter(X_train_ica[:, 0], X_train_ica[:, 1], c=y_train)
plt.title('ICA')

plt.subplot(1, 3, 3)
plt.scatter(X_train_lda[:, 0], X_train_lda[:, 1], c=y_train)
plt.title('LDA')

plt.tight_layout()
plt.show()

# Avaliar o poder discriminativo
# Treinar um classificador (por exemplo, Regressão Logística) em cada conjunto de dados transformado
# e comparar as precisões.

# PCA
clf_pca = LogisticRegression()
clf_pca.fit(X_train_pca, y_train)
y_pred_pca = clf_pca.predict(X_test_pca)
accuracy_pca = accuracy_score(y_test, y_pred_pca)

# ICA
clf_ica = LogisticRegression()
clf_ica.fit(X_train_ica, y_train)
y_pred_ica = clf_ica.predict(X_test_ica)
accuracy_ica = accuracy_score(y_test, y_pred_ica)

# LDA
clf_lda = LogisticRegression()
clf_lda.fit(X_train_lda, y_train)
y_pred_lda = clf_lda.predict(X_test_lda)
accuracy_lda = accuracy_score(y_test, y_pred_lda)

# Variáveis originais
clf_original = LogisticRegression()
clf_original.fit(X_train_scaled, y_train)
y_pred_original = clf_original.predict(X_test_scaled)
accuracy_original = accuracy_score(y_test, y_pred_original)

# Imprimir resultados
print(f"Acurácia com PCA: {accuracy_pca}")
print(f"Acurácia com ICA: {accuracy_ica}")
print(f"Acurácia com LDA: {accuracy_lda}")
print(f"Acurácia com variáveis originais: {accuracy_original}")

# Conclusões:
# Compare as precisões obtidas com cada método e com as variáveis originais.
# Analise qual método teve o melhor desempenho na tarefa de classificação.
# Discuta as vantagens e desvantagens de cada método.
```

```python
# prompt: Estime o resultado de classifica ̧c ̃ao obtido se utilizarmos limites (patamares)
# de sele ̧c ̃ao no primeiro PCD que maximize o produto das eficiˆencias (PE),
# para isso, calcule as probabilidades de detec ̧c ̃ao de cada classe para diferentes
# valores dos patamares.

import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import roc_curve, auc

# Calcular as probabilidades de detecção para cada classe
y_scores_pca = clf_pca.predict_proba(X_test_pca)

# Calcular as curvas ROC para cada classe
fpr = dict()
tpr = dict()
roc_auc = dict()
for i in range(3):
    fpr[i], tpr[i], _ = roc_curve(y_test == i, y_scores_pca[:, i])
    roc_auc[i] = auc(fpr[i], tpr[i])

# Encontrar o limiar que maximiza o produto das eficiências (PE)
thresholds = np.linspace(0, 1, 100)
pe_scores = []
for threshold in thresholds:
    # Calcular as eficiências para cada classe
    efficiencies = []
    for i in range(3):
        true_positives = np.sum((y_test == i) & (y_scores_pca[:, i] >= threshold))
        total_positives = np.sum(y_test == i)
        efficiency = true_positives / total_positives if total_positives > 0 else 0
        efficiencies.append(efficiency)

    # Calcular o produto das eficiências
    pe_score = np.prod(efficiencies)
    pe_scores.append(pe_score)

# Encontrar o limiar que maximiza o PE
optimal_threshold = thresholds[np.argmax(pe_scores)]

# Calcular a precisão com o limiar ótimo
y_pred_pca_optimal = (y_scores_pca >= optimal_threshold).astype(int)
accuracy_pca_optimal = accuracy_score(y_test, y_pred_pca_optimal.argmax(axis=1))

print(f"Limiar ótimo para maximizar PE: {optimal_threshold}")
print(f"Acurácia com PCA e limiar ótimo: {accuracy_pca_optimal}")

# Plotar as curvas ROC
plt.figure()
for i in range(3):
    plt.plot(fpr[i], tpr[i], label='Classe %d (AUC = %0.2f)' % (i, roc_auc[i]))
plt.plot([0, 1], [0, 1], 'k--')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('Taxa de Falsos Positivos')
plt.ylabel('Taxa de Verdadeiros Positivos')
plt.title('Curvas ROC para PCA')
plt.legend(loc="lower right")
plt.show()
```
